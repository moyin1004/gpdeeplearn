```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

def swish(x):
    a, b = x.chunk(2, dim=-1)
    return a * F.silu(b)

x_range = torch.linspace(-10, 10,2100)
x_input = x_range.unsqueeze(-1).expand(-1, 2) # 100*2
# x_input = x_range
print(x_input)
with torch.no_grad():
    y_swiglue = swish(x_input)
print(y_swiglue)
plt.figure(figsize=(8, 5))
plt.plot(x_range.numpy(), y_swiglue[:, 0].numpy(), label='SwiGLU Output',
         color='#FF6F00', linewidth=2)
plt.xlabel('Input')
plt.ylabel('Output')
plt.legend()
plt.show()
```
